**Import libraries**

```python
import requests
import json  # optional, for pretty printing later
```

**Set API URL & parameters**

```python
url = "https://api.coingecko.com/api/v3/coins/markets"

params = {
    "vs_currency": "usd",       # currency
    "ids": "bitcoin,ethereum"   # coins we want
}
```

**Call the API**

```python
response = requests.get(url, params=params)
```

**Convert to JSON**

```python
data = response.json()
data
```

**Print**

```python
print(json.dumps(data, indent=2))
```
